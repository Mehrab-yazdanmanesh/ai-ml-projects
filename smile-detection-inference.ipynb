# Smile Detection — Inference

Place the model checkpoint in this folder and set `address` to an image file.

```python
from keras.models import load_model
import cv2 as cv
import numpy as np
```

```python
model=load_model("Smile1_model.keras", compile=False)
```

```python
address = "sample.jpg"
```

```python
img = cv.imread(address)
if img is None:
    raise FileNotFoundError(f"Cannot read {address}")
f_img=cv.bilateralFilter(img,d=9,sigmaColor=75,sigmaSpace=75)
std_dev=np.std(f_img)
print(std_dev)
if std_dev>=0 and std_dev<=50:
    alpha=1.5
elif std_dev>=50 and std_dev<=100:
    alpha=1.0
elif std_dev>=100 and std_dev<=150:
    alpha=0.75
elif std_dev>150:
    alpha=0.50
reso_img=cv.convertScaleAbs(f_img,alpha=alpha,beta=0)
lab=cv.cvtColor(reso_img,cv.COLOR_RGB2Lab)
l,a,b=cv.split(lab)
image=cv.createCLAHE(clipLimit=2.0,tileGridSize=(8,8))
l=image.apply(l)
lab=cv.merge((l,a,b))
n_img=cv.cvtColor(lab,cv.COLOR_Lab2RGB)
r_img=cv.resize(n_img,(25,25))
img=cv.normalize(r_img,None,0,255,cv.NORM_MINMAX)
```

```python
test=model.predict(np.expand_dims(img,axis=0))
non_smile =test[0][0]
smile =test[0][1]
if non_smile > smile:
    print("smile")
elif non_smile < smile:
    print("non_smile")
```
