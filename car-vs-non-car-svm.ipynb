```python
import numpy as np
import cv2 as cv
import glob
from sklearn.model_selection import train_test_split
from sklearn import svm
import joblib as jb
from sklearn.metrics import accuracy_score
import cv2 as cv
import glob
import numpy as np
from sklearn import svm
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
from keras.utils import to_categorical
from keras import layers,models
from sklearn.metrics import precision_score,accuracy_score,recall_score,f1_score,roc_auc_score
import tensorflow as tf
from keras import utils
from keras.models import load_model
from keras import layers,models
from keras.optimizers import Adam

```

```python
pic=[]
labels=[]

```

```python
from pathlib import Path

dataset_dir = Path("data/car-vs-non-car")
```

```python
for i in sorted(dataset_dir.glob("*/*")):
    img = cv.imread(str(i)) if i.is_file() else None
    if img is None:
        continue
    r_img=cv.resize(img,(25,25)).flatten()
    pic.append(r_img)
    label=i.parent.name
    labels.append(label)
```

```python
print(labels)
```

```python
model=svm.SVC(kernel="linear")
x_train,x_test,y_train,y_test=train_test_split(pic,labels,test_size=0.05,random_state= 5)
model.fit(x_train,y_train)
print("accuracy : %.2f"%(model.score(x_test,y_test)))
```

```python
sample_path = "sample.jpg"
```

```python
img = cv.imread(sample_path)
if img is None:
    raise FileNotFoundError(f"Cannot read {sample_path}")
features = cv.resize(img, (25, 25)).reshape(1, -1)
print(model.predict(features)[0])
```
