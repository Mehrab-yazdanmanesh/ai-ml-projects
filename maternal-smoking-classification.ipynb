# Maternal Smoking Status Classification

Place the input CSV at `data/babies.csv`.

```python
import pandas as pd 
import cv2 as cv
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score,roc_auc_score
from sklearn.preprocessing import OneHotEncoder
from keras import models,layers
from keras import utils
import tensorflow as tf
from sklearn.compose import ColumnTransformer
from keras.optimizers import Adam
```

```python
df=pd.read_csv("data/babies.csv")
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
df["gestation"].value_counts()
df["height"].value_counts()
df["weight"].value_counts()
df["age"] = df["age"].fillna(df["age"].mean())
df["smoke"] = df["smoke"].fillna(df["smoke"].mean())
```

```python
df["gestation"] = df["gestation"].fillna(282.0)
df["height"] = df["height"].fillna(64.0)
df["weight"] = df["weight"].fillna(130.0)
```

```python
one=OneHotEncoder()
```

```python
x=df.drop("smoke",axis=1)
y=df["smoke"]
```

```python
df.isna().sum()
```

```python
categori=["case","gestation"]
```

```python
transform=ColumnTransformer([("mmd_code",one,categori)],remainder="passthrough")
```

```python
transform
```

```python
transformer_x=transform.fit_transform(x)
```

```python
transformer_x=pd.DataFrame(transformer_x)
```

```python
transformer_x
```

```python
y=utils.to_categorical(y)
print(y)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.01,random_state=5)
```

```python
net=models.Sequential([layers.Dense(100,activation="relu",input_dim=(7)),
                        layers.Dense(50,activation="relu"),
                        layers.Dense(2,activation="sigmoid")])
```

```python
net.summary()
```

```python
op=Adam(learning_rate=0.0001)
net.compile(optimizer=op,loss="binary_crossentropy",metrics=["accuracy"])
```

```python
net.fit(x_train,y_train,epochs=10,batch_size=5,validation_data=(x_test,y_test))
```

```python
y_pred=net.predict(x_test).round()
```

```python
y_test=tf.convert_to_tensor(y_test)
y_pred=tf.convert_to_tensor(y_pred)
y_test=y_test.numpy().ravel()
y_pred=y_pred.numpy().ravel()
```

```python
accuracy=accuracy_score(y_test,y_pred)
precision=precision_score(y_test,y_pred)
recall=recall_score(y_test,y_pred)
f1=f1_score(y_test,y_pred)
```

```python
print("%.2f"%(accuracy*100))
print("%.2f"%(precision*100))
print("%.2f"%(recall*100))
print("%.2f"%(f1*100))
```

```python
net.save("babies.keras")
```
