# Car vs. Non-Car — Dense Network

Place class folders directly inside `data/car-vs-non-car/`, or change `dataset_dir`.

```python
import cv2 as cv
import glob
import numpy as np
from sklearn import svm
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
from keras.utils import to_categorical
from keras import layers,models
from sklearn.metrics import precision_score,accuracy_score,recall_score,f1_score,roc_auc_score
import tensorflow as tf
from keras import utils
```

```python
pic=[]
labels=[]
```

```python
from pathlib import Path

dataset_dir = Path("data/car-vs-non-car")
```

```python
for i in sorted(dataset_dir.glob("*/*")):
    img = cv.imread(str(i)) if i.is_file() else None
    if img is None:
        continue
    r_img=cv.resize(img,(25,25)).flatten()
    pic.append(r_img)
    label=i.parent.name
    labels.append(label)
```

```python
pic=np.array(pic)
```

```python
print(labels)
```

```python
le=LabelEncoder()           
```

```python
labels=le.fit_transform(labels)
labels=utils.to_categorical(labels)
```

```python
pic=np.array(pic)
```

```python
print(labels)
```

```python
x_train,x_test,y_train,y_test=train_test_split(pic,labels,train_size=0.98)
```

```python
net = models.Sequential([layers.Dense(500,activation="relu",input_dim=(1875)),
                         layers.Dense(450,activation="relu"),
                         layers.Dense(400,activation="relu"),
                         layers.Dense(350,activation="relu"),
                         layers.Dense(300,activation="relu"),
                         layers.Dense(250,activation="relu"),
                         #layers.Dense(20,activation="relu"),
                         #layers.Dense(20,activation="relu"),
                         #layers.Dense(10,activation="relu"),       
                         layers.Dense(2,activation="sigmoid")])
```

```python
net.summary()
```

```python
net.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy"])
```

```python
net.fit(x_train,y_train,epochs=1,batch_size=100,validation_data=(x_test,y_test))
```
